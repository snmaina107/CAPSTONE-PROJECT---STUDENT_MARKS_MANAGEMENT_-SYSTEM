## Problem Description

In any school or university,lecturers must keep track of their students' performance. At the end of each assessment,a lecturer ends up with a pile of records: each students' ID,Name,Course and Marks.Doing this on paper or in a loose spreadsheet quickly becomes unmanageable,especially in a class of dozens or hundreds of students.
The lecturer needs to do several things with all this data all the time:

•	Register a new student who joins late.

•	Correct a mark after a recheck or a marking error.

•	Remove a student who has withdrawn from the course.

•	Look up one student's record quickly using their ID.

•	Rank students from highest to lowest marks.

•	Find the top scorer, the lowest scorer and the class average.

Mistakes are easy to make.Two students can be given the same ID,marks outside 0-100 range can be typed in by accident and a search for a student who is not registered can go unnoticed.The records change constantly,so the system has to handle additions,corrections and removals of student records without breaking.

**Why this requires a data structure and an algorithmic solution**

A data structure is needed because the records must be stored in an organized way that supports frequent changes.The number of students is not fixed.It grows when students join and shrinks when they leave. A fixed array handles this poorly,because deleting and inserting in the middle means shifting every element after it.
A singly linked list fit better: each student is stored in a node that point to the next one,so adding or removing a record only requires changing a few links.Nothing needs to be shifted.
Algorithms are needed too because storing data is only half the problem.The lecturer also needs answers from it and getting them takes a clear step-by-step method:

•	Linear search finds a student by ID. A linked list has no index, so we must check the nodes one at a time from the head.

•	Bubble sort and selection sort rank students by marks, so we can compare two different sorting approaches.

•	Traversal calculates the highest mark, lowest mark, class average and number of students.

The choice of data structure and algorithm also affects efficiency. For example, search takes O(n) time on a linked listed and the simple sorts take O(n^2). Analyzing this shows how te solution behaves as the class grows.

**The intended purpose of the solution**

The purpose of this project is to build a Student Marks Management System: a menu-driven python program that stores student records in a singly linked list implemented from scratch. It lets the lecturer add,display,search,update and delete students,sort them by marks and view class statistics.It also handles errors gracefully:an empty list,student not found ,a duplicate ID, invalid marks and invalid menu choices.
The project also serves a learning purpose.By building the linked list ourselves instead of using Python's built-in list,we show how nodes and reference work in memory,apply searching and sorting algorithms and analyze their time complexity.


##  Data Structure Implementation

We used a **Singly Linked List**. Each **node** stores one student's record plus a reference (next) to the next node.


* **Head** points to the first node. If the list is empty the Head is None.
* The last node's next is **None**.
* To reach any student we must start at head and follow the next references using the transversal concept.

**Why a linked list?** Adding at the end/beginning and deleting a node only needs reference changes, with no shifting of elements as in an array.

# The  creation of the Node class
One node = one student record + a link to the next node.

````python
class Node:
    # This block of code creates a single node of the linked list that holds one student's record.
    def __init__(self, student_id, name, course, marks):
        self.student_id = student_id
        self.name = name
        self.course = course
        self.marks = marks
        self.next = None          # reference to the next node (None = end of list)
````

###  The StudentLinkedList class

This class keeps track of the `head` of the list.

We start with a **small class** that only has the basics: the `head`, `is_empty()` and `count_students()`.

Then, in each section below, we wrote one operation as its own function and **attached it to the class** with a line like `StudentLinkedList.add_student = add_student`. This way every operation sat in its own cell, right next to its explanation.


````python
# The class only keeps track of the head (the first node).
# Each operation is added in its own section below.
class StudentLinkedList:
    """Singly Linked List that stores student records."""

    def __init__(self):
        self.head = None                       # empty list

    def is_empty(self):
        """True if the list has no nodes."""
        return self.head is None

    def count_students(self):
        count = 0
        current = self.head
        while current is not None:
            count += 1
            current = current.next
        return count
````

## The ADD Operation

`add_student()` creates a new **Node** and links it at the **end** of the list plus it first rejects **duplicate IDs** and **invalid marks**.
The operation adds more data in what you alreadyy have.
Let's load all 12 sample records, then try to add a duplicate and an invalid mark.

````python
# Linear search. 
# It is written here first because add_student uses it to reject duplicate IDs as it goes through the list once to find the end of the list.

def search_by_id(self, student_id):
    """Linear search: check nodes one by one from head to tail.
    Returns the node if found, otherwise None."""
    current = self.head
    while current is not None:
        if current.student_id == student_id:
            return current
        current = current.next
    return None

StudentLinkedList.search_by_id = search_by_id
````

````python
#  ADD  (append at the end)            Time: O(n)
def add_student(self, student_id, name, course, marks):
    
    # error handling: check for duplicate ID and valid marks.
    
    if self.search_by_id(student_id) is not None:
        print(f" Error: Student ID '{student_id}' already exists (duplicate ID).")
        return False
    if not isinstance(marks, (int, float)) or not (0 <= marks <= 100):
        print(" Error: Marks must be a number between 0 and 100.")
        return False

    new_node = Node(student_id, name, course, marks)
    if self.head is None:                  # empty list -> new node is the head
        self.head = new_node
    else:                                  # otherwise goto the last node.
        current = self.head
        while current.next is not None:
            current = current.next
        current.next = new_node            # link the last node to the new node
    print(f"Added: {student_id} - {name}")
    return True

StudentLinkedList.add_student = add_student
````

## The Display Operation

`display_all()` **traverses** the list from `head` and prints each node. If the list is empty it prints a warning instead.

````python
# DISPLAY ALL                         Time: O(n)
def display_all(self):
    if self.is_empty():
        print("The list is empty. No students to display.")
        return
    print(f"{'ID':<8}{'Name':<22}{'Course':<12}{'Marks':>6}")
    print("-" * 48)
    current = self.head
    while current is not None:
        print(f"{current.student_id:<8}{current.name:<22}{current.course:<12}{current.marks:>6}")
        current = current.next

StudentLinkedList.display_all = display_all
````

## Search Operation (Linear Search)

**Linear Search** checks each node one after the other, starting from the head, until it finds the matching ID or reaches the end (`None`). Off course the Best case is when the student is the first node → **O(1)** but the Worst case is when the student is last or does not exist **O(n)**

**The Linear Search code** (written at the start of Section add operation, because `add_student` used it to check for duplicate IDs):

```python
#  SEARCH BY ID  (LINEAR SEARCH)       Time: O(n)
def search_by_id(self, student_id):
    """Linear search: check nodes one by one from head to tail.
    Returns the node if found, otherwise None."""
    current = self.head
    while current is not None:
        if current.student_id == student_id:
            return current
        current = current.next
    return None
```

**How it works:** it starts at the `head` and checks each node. If the ID matches, it returns that node. If not, it moves to `next`. If it reaches `None`, the student does not exist, so it returns `None`.

`search_and_show` code below calls it and prints a friendly  ouput message after executing it.

````python
def search_and_show(self, student_id):
    
    """Search and print a friendly result (used by the menu)."""
    
    if self.is_empty():
        print("The list is empty. Nothing to search.")
        return None
    node = self.search_by_id(student_id)
    if node is None:
        print(f"Student with ID '{student_id}' not found.")
    else:
        print(f"Found: ID={node.student_id}, Name={node.name}, "
              f"Course={node.course}, Marks={node.marks}")
    return node

StudentLinkedList.search_and_show = search_and_show
````

## Update Operation

`update_marks()` uses the search to locate the node and then changes its `marks`.

````python
#  UPDATE MARKS                        Time: O(n)
def update_marks(self, student_id, new_marks):
    if self.is_empty():
        print("The list is empty. Nothing to update.")
        return False
    if not isinstance(new_marks, (int, float)) or not (0 <= new_marks <= 100):
        print("Error: Marks must be a number between 0 and 100.")
        return False
    node = self.search_by_id(student_id)
    if node is None:
        print(f"Student with ID '{student_id}' not found.")
        return False
    old = node.marks
    node.marks = new_marks
    print(f"Updated {node.name}: {old} → {new_marks}")
    return True

StudentLinkedList.update_marks = update_marks
````

##  Delete Operation

Deleting a node means **bypassing** it by changing links. We are removing the data from the record completely.
Though there are two cases:

```text
Case 1 ->delete the head:       head ->[A]->[B]->[C]      becomes   head->[B]->[C]
Case 2 ->delete a later node:   [A]->[B]->[C]  (delete B) it becomes   [A]->[C]
```

````python
#  DELETE                             Time: O(n)
def delete_student(self, student_id):
    if self.is_empty():
        print("The list is empty. Nothing to delete.")
        return False

    # Case 1: the student to delete is the head node
    if self.head.student_id == student_id:
        print(f"Deleted: {self.head.student_id} - {self.head.name}")
        self.head = self.head.next
        return True

    # Case 2: the student is somewhere after the head
    previous = self.head
    current = self.head.next
    while current is not None:
        if current.student_id == student_id:
            previous.next = current.next   # skip over (bypass) the node
            print(f"Deleted: {current.student_id} - {current.name}")
            return True
        previous = current
        current = current.next

    print(f"Student with ID '{student_id}' not found.")
    return False

StudentLinkedList.delete_student = delete_student
````

##  Sorting Operations

Both algorithms sort the students **by marks**. They work by swapping the data held in two nodes (the links themselves are not changed), and both accept `ascending=True/False`.

###  Bubble Sort
Repeatedly compares **neighbouring** nodes and swaps them if they are in the wrong order. After each pass the largest remaining mark "bubbles" to the end. If a full pass makes **no swaps**, the list is already sorted and we stop early.

````python
def swap_data(a, b):
    
    """Swap the record stored in two nodes (the links stay untouched)."""
    
    a.student_id, b.student_id = b.student_id, a.student_id
    a.name,       b.name       = b.name,       a.name
    a.course,     b.course     = b.course,     a.course
    a.marks,      b.marks      = b.marks,      a.marks


#  BUBBLE SORT (by marks)            Time: O(n^2)
def bubble_sort(self, ascending=True):
    
    """Repeatedly compare neighbouring nodes and swap them if out of order."""
    
    if self.is_empty():
        print("The list is empty. Nothing to sort.")
        return
    end = None                             
    # nodes from 'end' onwards are already sorted
    while True:
        swapped = False
        current = self.head
        while current.next is not end:
            out_of_order = (current.marks > current.next.marks) if ascending \
                           else (current.marks < current.next.marks)
            if out_of_order:
                swap_data(current, current.next)
                swapped = True
            current = current.next
        end = current                      # the last node is now in its final place
        if not swapped:                    # no swaps -> list is sorted (early exit)
            break
    order = "ascending" if ascending else "descending"
    print(f"Sorted by marks ({order}) using Bubble Sort.")

StudentLinkedList.bubble_sort = bubble_sort
````

````python
print("BEFORE sorting:")
students.display_all()

print()
students.bubble_sort(ascending=True)
students.display_all()

print()
students.bubble_sort(ascending=False)
students.display_all()
````

### Selection Sort
For each position, **selects** the smallest or largest, (depending on the order we want-Ascending or Descending) remaining node and swaps it into place. It makes fewer swaps than bubble sort, but always performs about n²/2 comparisons.

````python
# SELECTION SORT (by marks)         Time: O(n^2)
def selection_sort(self, ascending=True):
    """For each position, find the smallest (or largest) remaining node and
    swap it into place."""
    if self.is_empty():
        print("The list is empty. Nothing to sort.")
        return
    current = self.head
    while current is not None:
        best = current                     # node with the best mark found so far
        scanner = current.next
        while scanner is not None:
            better = (scanner.marks < best.marks) if ascending else (scanner.marks > best.marks)
            if better:
                best = scanner
            scanner = scanner.next
        if best is not current:
            swap_data(current, best)
        current = current.next
    order = "ascending" if ascending else "descending"
    print(f"Sorted by marks ({order}) using Selection Sort.")

StudentLinkedList.selection_sort = selection_sort
````

````python
print("BEFORE sorting:")
students.display_all()

print()
students.selection_sort(ascending=True)
students.display_all()

print()
students.selection_sort(ascending=False)
students.display_all()
````

### Highest, Lowest, Average and Count
Each of these traverses the whole list once → **O(n)**.

````python
# HIGHEST / LOWEST / AVERAGE / COUNT         Time: O(n) each
def highest_mark(self):
    if self.is_empty():
        print("The list is empty.")
        return None
    best = self.head
    current = self.head.next
    while current is not None:
        if current.marks > best.marks:
            best = current
        current = current.next
    return best                            # returns the node (so we know who scored it)

StudentLinkedList.highest_mark = highest_mark


def lowest_mark(self):
    if self.is_empty():
        print("The list is empty.")
        return None
    worst = self.head
    current = self.head.next
    while current is not None:
        if current.marks < worst.marks:
            worst = current
        current = current.next
    return worst

StudentLinkedList.lowest_mark = lowest_mark


def class_average(self):
    if self.is_empty():
        print("The list is empty.")
        return None
    total, count = 0, 0
    current = self.head
    while current is not None:
        total += current.marks
        count += 1
        current = current.next
    return total / count

StudentLinkedList.class_average = class_average
````

````python
top = students.highest_mark()
low = students.lowest_mark()
print(f"Highest mark : {top.marks}  ({top.name})")
print(f"Lowest mark  : {low.marks}  ({low.name})")
print(f"Class average: {students.class_average():.2f}")
print(f"Student count: {students.count_students()}")
````

## Menu-Driven Program

The menu lets the lecturer choose operations repeatedly until they choose **Exit** (12). It handles **invalid menu choices** and **invalid number input** without crashing.

It uses Python's `input()`, so the notebook waits for you to type a choice and press Enter.

````python
import io, contextlib     

def print_menu():
    print("\n" + "=" * 42)
    print("   STUDENT MARKS MANAGEMENT SYSTEM")
    print("=" * 42)
    print(" 1. Add a student")
    print(" 2. Display all students")
    print(" 3. Search for a student by ID")
    print(" 4. Update student marks")
    print(" 5. Delete a student")
    print(" 6. Sort by marks (Bubble Sort)")
    print(" 7. Sort by marks (Selection Sort)")
    print(" 8. Find the highest mark")
    print(" 9. Find the lowest mark")
    print("10. Calculate the class average")
    print("11. Count students")
    print("12. Exit")
    print("-" * 42)


def read_marks():
    """Ask for marks and make sure a number is entered."""
    try:
        value = float(input("Enter marks (0-100): "))
        return int(value) if value.is_integer() else value
    except ValueError:
        print("Error: Marks must be a number.")
        return None


def run_menu(student_list):
    while True:
        print_menu()
        choice = input("Enter your choice (1-12): ").strip()

        if choice == "1":
            sid    = input("Enter student ID: ").strip()
            name   = input("Enter name: ").strip()
            course = input("Enter course: ").strip()
            marks  = read_marks()
            if marks is not None:
                student_list.add_student(sid, name, course, marks)

        elif choice == "2":
            student_list.display_all()

        elif choice == "3":
            student_list.search_and_show(input("Enter student ID to search: ").strip())

        elif choice == "4":
            sid   = input("Enter student ID: ").strip()
            marks = read_marks()
            if marks is not None:
                student_list.update_marks(sid, marks)

        elif choice == "5":
            student_list.delete_student(input("Enter student ID to delete: ").strip())

        elif choice == "6":
            student_list.bubble_sort(ascending=True)
            student_list.display_all()

        elif choice == "7":
            student_list.selection_sort(ascending=True)
            student_list.display_all()

        elif choice == "8":
            node = student_list.highest_mark()
            if node: print(f"Highest mark: {node.marks} ({node.name})")

        elif choice == "9":
            node = student_list.lowest_mark()
            if node: print(f"Lowest mark: {node.marks} ({node.name})")

        elif choice == "10":
            avg = student_list.class_average()
            if avg is not None: print(f"Class average: {avg:.2f}")

        elif choice == "11":
            print(f"Number of students: {student_list.count_students()}")

        elif choice == "12":
            print("Goodbye!")
            break

        else:
            print("Invalid choice. Please enter a number from 1 to 12.")
````

## Complexity Analysis of this project

Let **n** = number of students in the linked list.

**Time complexity**:

**Add Operation (append at end)** has a time complexity of **O(n)** - We must traverse to the last node because we keep no `tail` reference. (The duplicate-ID check is also O(n).) With a tail pointer it could be O(1) for the link itself.

**Display all** has a time complexity of **O(n)** -It Visits every node once.

**Search by ID (Linear Search)**,Its best case has a time complexity of **O(1)**, Worst/Average case is **O(n)** .How is this so: Best case: ID is at the head. Worst case: ID is last or missing.

**Update marks** has a time complexity of **O(n)** - Needs a linear search first(it has to go through the data one by one at a time); changing the value itself is O(1).

**Delete** Best case has a time complexity of **O(1)**, Worst case is **O(n)** Deleting the head is O(1); otherwise we traverse to find the previous node.

**Bubble Sort** Best case has a time complexity of **O(n)**, Average/Worst **O(n²)** - Best case (already sorted) is O(n) because of the early-exit `swapped` check(we have to confirm so it will still go through the linked list once checkng every entry).

**Selection Sort** is **O(n²)** (best, average and worst) because it always scans the remaining nodes for each position, even if already sorted. 

**Highest / Lowest mark** is **O(n)** - Must compare every node.

**Class average** is **O(n)** - Sum all marks in one traversal. 
**Count students** is **O(n)** - As it traverses the list (could be O(1) if we stored a `size` counter).

**Space complexity:**
* The linked list uses **O(n)** space (one node per student).
* Bubble sort and selection sort sort **in place**, swapping data between nodes, so they use only **O(1)** extra space.

**Comparison with arrays:** a Python list gives O(1) access by index, but inserting/deleting in the middle will need shifting. A linked list has O(1) insert/delete *once the node is found*, but finding a node always needs traversal.


## Challenges

1. **Deleting nodes correctly** – deleting the head is different from deleting a middle or last node, so we needed two separate cases and a `previous` pointer.
2. **Sorting a linked list** – we cannot access nodes by index like in an array, so the sorting algorithms had to be written using node references. We solved this by swapping the *data* between nodes instead of re-linking them.
3. **Handling errors** – we had to think about an empty list, duplicate IDs, marks outside 0–100, a student that does not exist, and wrong menu input.
4. **Menu input in Jupyter** – `input()` pauses the notebook, so the notebook waits until a choice is typed, and we had to handle wrong input such as letters instead of marks.
